```python
import pandas as pd

data = {
    "Name": ["Aman", "Riya", None, "Vikas"],
    "Dept": ["Sales", None, "HR", "IT"],
    "Age": [28, 32, 25, None],
    "Bonus": [None, 5000, 3000, 7000]
}

df = pd.DataFrame(data)
print(df)
print("Original shape:", df.shape)
```

```python
# Step 1: dropna with subset (only checks Name and Dept)
df_clean = df.dropna(subset=["Name", "Dept"])
print(df_clean)
print("After subset dropna:", df_clean.shape)
```

```python
# Step 2: dropna with no subset (checks every column)
df_all = df.dropna()
print(df_all)
print("After full dropna:", df_all.shape)
```

```python

print("Step 0 - original:", df.shape)              # (4, 4)
print("Step 1 - subset dropna:", df_clean.shape)    # (2, 4)
print("Step 2 - full dropna:", df_all.shape)         # (0, 4)
```
