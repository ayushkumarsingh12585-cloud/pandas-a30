```python
import pandas as pd
import numpy as np

data = {
    "id": [1, 2, 2, 3, 4, 4],
    "age": [25, np.nan, 30, 30, np.nan, np.nan],
    "city": ["Delhi", "Mumbai", "Mumbai", np.nan, "Jaipur", "Jaipur"]
}

df = pd.DataFrame(data)
print(df)
print("Shape:", df.shape)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
print(df.duplicated())
```

```python
print(df.duplicated().sum())
print(df.shape)
```

```python
df["age"] = df["age"].fillna(df["age"].mean())
print(df)
print("Shape:", df.shape)
```

```python
df["city"] = df["city"].fillna(df["city"].mode())
print(df)
print(df.shape)
```

```python
df = df.drop_duplicates().reset_index(drop=True)
print(df)
print("Shape:", df.shape)
```

```python
print("Null values:\n", df.isnull().sum())
print("Duplicate rows:", df.duplicated().sum())
print(df.info())
print("\n", df.describe())
```
