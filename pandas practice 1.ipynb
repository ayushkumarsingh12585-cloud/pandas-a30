```python
import pandas as pd

data = {
    "name": ["virat", "mohit", "rohit", "mohit"],
    "age": [25, 31, 30, 31],
    "city": ["jaipur", "gurgaon", "rewari", "gurgaon"]
}

df = pd.DataFrame(data)

print(df)
```
