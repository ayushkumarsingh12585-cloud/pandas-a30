```python
import pandas as pd

data = {
    "Name": ["Aman", "Riya", "Vikas"],
    "Dept": ["Sales", "HR", "IT"],
    "Age": [28, 32, 25]
}

df = pd.DataFrame(data)
print("Before:")
print(df)

# Rename specific columns
df = df.rename(columns={"Name": "Employee_Name", "Dept": "Department"})

# Convert all column names to lowercase
df_lower = df.rename(columns=str.lower)

# Convert all column names to uppercase
df_upper = df.rename(columns=str.upper)

print("\nAfter rename:")
print(df)

print("\nLowercase columns:")
print(df_lower)

print("\nUppercase columns:")
print(df_upper)
```

```python
import pandas as pd

data = {"Amount": [100, 120, 110, 130, 5000]}  # 5000 is an outlier
df = pd.DataFrame(data)
print(df)

Q1 = df["Amount"].quantile(0.25)
Q3 = df["Amount"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[(df["Amount"] < lower) | (df["Amount"] > upper)]
print("Outliers:\n", outliers)

clean_df = df[(df["Amount"] >= lower) & (df["Amount"] <= upper)]
print("Cleaned:\n", clean_df)
```
